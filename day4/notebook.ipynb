## 1. Quick theory: MDP and Q-learning

- MDP: (S, A, P, R, gamma). We will use deterministic transitions.
- Q-learning update: Q(s,a) <- Q(s,a) + alpha [r + gamma max_a' Q(s',a') - Q(s,a)]
- Epsilon-greedy exploration for action selection.
