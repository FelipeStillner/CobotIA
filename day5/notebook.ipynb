## 3 — URScript via TCP (send script commands)

You can also open a TCP socket to port 30002 and send URScript commands.
Use this with extreme caution and only for small tested commands.


```python
# Student Task: send a small URScript command to the robot via socket (do NOT run without permission)
import socket

def send_urscript(robot_ip, port, script):
    s = socket.socket()
    s.connect((robot_ip, port))
    s.send((script+'\n').encode('utf8'))
    
print('Define send_urscript but DO NOT execute on the real robot without instructor OK')
```
